### **Problem:**
     
     You are given ab array A of n elements.
     
     You have to remove exactly n/2 elements from an arrat and add it to another array B(initially emplty).
     
     Find the maximum and minimum values of difference between these two arrays.

     The difference between those two arrays is sum(abs(A[i]-B[j])).

POINTS TO REMEMBER:-

---


       
*   Local Best solution is the Global Best Solution.
*   Maximum condition:-

        Σabs(a[i] - a[j])        
              i = i + 1             
              j = j + 1

        Here , the loop will run 'n/2' times.

*   Minimum condition:-

        Σabs(|a[2n] - a[2n+1]|)

        Here the loop will run only n/2 times because of '2n'.





```python
def max_min_Diff(arr):
  arr.sort()
  n = len(arr)
  mid = n//2
  max = 0
  min = 0
  j = n-1

  for i in range(mid):
    max += abs(arr[i] - arr[j])
    j -= 1

    min += abs(arr[2*i] - arr[2*i+1])

  print(f"Maximum Difference is {max}")
  print(f"Minimum Difference is {min}")

arr = []
n = int(input("Enter the length of array : "))
for i in range(0,n):
  arr.append(int(input("Enter the element : ")))
print(arr)

max_min_Diff(arr)
```

## **COIN CHANGE PROBLEM**

---



### Ques-2. Find the minimum number of denominations.

```python
def min_coins(coins,amount):
  result = []
  for coin in coins:
    while amount >= coin:
      amount -= coin     #coins ki value ko amount se minus kerna
      result.append(coin)     #result list me coin ki value ko add kerna

  return result
#greedy approach se sbse bada coin pehle use kerte h isiliye descending order me coins ki list h
coins = [500, 100 , 50 , 20, 10,5,2,1]
amount = int(input("Enter the amount : "))
result = min_coins(coins,amount)
print(result)
```
