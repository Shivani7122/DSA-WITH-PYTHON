***DOUBLY LINKED LIST***

```python
class Node:
  def __init__(self,value=None):
    self.data = value
    self.next = None
    self.prev = None

class DoublyLL:
  def __init__(self):
    self.head = None

 #Insertion At the mid
  def insert_at_mid(self,value,x):
    t = self.head
    while t.next is not None:
      if t.data == x:
        break
      else:
        t = t.next
    temp = Node(value)
    temp.next = t.next
    t.next.prev = temp
    t.next = temp
    temp.prev = t

  #Insertion At the Beginning
  def insert_at_beginning(self,value):
    temp = Node(value)
    if self.head is None:
      self.head = temp
      return
    temp.next = self.head
    self.head.prev = temp
    self.head = temp


  #Insertion At the End
  def insert_at_end(self,value):
    temp = Node(value)
    if self.head is None:
      self.head = temp
      return
    t = self.head
    while t.next is not None:
      t = t.next
    t.next = temp
    temp.prev = t
  #Print DLL
  def printDLL(self):
    t = self.head
    while t.next is not None:
      print(t.data,end=" <---> ")
      t = t.next
    print(t.data)

  #Deletion
  def deleteDLL(self,value):
    #DLL is Empty
    if self.head is None:
      print("Linked List is Empty")
      return
    t = self.head
    #delete the first element in DLL
    if t.data == value:
      self.head = t.next
      self.head.prev = None
      return
    #delete the mid elements in DLL
    while t.next is not None:
      if t.data == value:
        t.prev.next = t.next
        t.next.prev = t.prev
        return
      t = t.next
    #delete the last element in DLL
    if t.data == value:
      t.prev.next = None

obj = DoublyLL()
obj.insert_at_end(60)
obj.insert_at_end(70)
obj.insert_at_end(80)
obj.insert_at_end(90)
obj.insert_at_end(100)
obj.insert_at_beginning(40)
obj.insert_at_beginning(30)
obj.insert_at_beginning(20)
obj.insert_at_beginning(10)
obj.insert_at_mid(50,40)
obj.deleteDLL(100)
obj.printDLL()


```

```python
class Node:
  def __init__(self,value=None):
    self.data = value
    self.next = None
    self.prev = None

class Cll:
  def __init__(self):
    self.head = None

  #Insertion at end
  def insertAtEnd(self,value):
    temp = Node(value)
    if self.head is None:
      self.head = temp
      temp.next = self.head   #Points to itself to start the circle
    else:
      t = self.head
      while t.next is not self.head:
        t = t.next
      t.next = temp
      temp.next  = self.head

  #Insertion At beginning
  def insertAtBeg(self,value):
    temp = Node(value)
    if self.head is None:
      self.head = temp
      temp.next = self.head
    else:
      t = self.head
      while t.next is not self.head:
        t = t.next
      t.next = temp
      temp.next = self.head
      self.head = temp

      #Insertion At Mid
  def insertAtMid(self,value,x):
    temp = Node(value)
    t = self.head
    while t.next is not self.head:
      if  t.data == x:
        break
      else:
        t = t.next
    temp.next = t.next
    t.next  = temp


  def deleteNode(self, x):
    # Case 0: Agar list khali hai
    if self.head is None:
        print("List empty hai!")
        return

    t = self.head
    prev = None

    # Case 1: Agar HEAD node hi delete karna ho
    if t.data == x:
        # Agar sirf ek hi node hai list mein
        if t.next == self.head:
            self.head = None
        else:
            # Last node dhundna padega taaki uska next update ho sake
            while t.next is not self.head:
                t = t.next
            t.next = self.head.next
            self.head = self.head.next
        return

    # Case 2: Agar node beech mein ya end mein ho
    t = self.head
    while t.next is not self.head:
        prev = t
        t = t.next
        if t.data == x:
            prev.next = t.next
            return

    # Agar poori list ghum li aur value nahi mili
    print(f"Value {x} doesn't exist.")



# Print CLL
  def printCLL(self):
      if self.head is None:
          print("List is empty")
          return

      t = self.head
      while True:
          print(t.data, end=" -> ")
          t = t.next
          if t == self.head: # Break once we've looped back to the start
              break
      print("(Back to Head)")

obj = Cll()
obj.insertAtBeg(20)
obj.insertAtBeg(10)
obj.insertAtMid(30,20)
obj.insertAtEnd(40)
obj.deleteNode(40)
obj.printCLL()

```

CLL FROM DLL

```python
class Node:
    def __init__(self, value=None):
        self.data = value
        self.next = None
        self.prev = None

class DCll:
    def __init__(self):
        self.head = None

    # 1. Insertion at End
    def insert(self, value):
        new_node = Node(value)
        if not self.head:
            self.head = new_node
            new_node.next = self.head
            new_node.prev = self.head
        else:
            last = self.head.prev # DCLL mein last node hamesha head.prev hota hai

            last.next = new_node
            new_node.prev = last
            new_node.next = self.head
            self.head.prev = new_node

    # 2. Deletion by Value
    def delete(self, key):
        if not self.head:
            print("List empty hai!")
            return

        curr = self.head

        # Searching for the node
        while True:
            if curr.data == key:
                # Case: Agar sirf ek hi node hai
                if curr.next == self.head and curr.prev == self.head:
                    self.head = None
                else:
                    # Pointers update karein
                    curr.prev.next = curr.next
                    curr.next.prev = curr.prev

                    # Agar head node delete ho raha ho
                    if curr == self.head:
                        self.head = curr.next
                return

            curr = curr.next
            if curr == self.head: # Poora circle ghoom liye
                break

        print(f"Value {key} nahi mili!")

    # 3. Printing (Forward and Backward check)
    def display(self):
        if not self.head:
            print("List empty hai!")
            return

        temp = self.head
        print("Forward: ", end="")
        while True:
            print(temp.data, end=" <-> ")
            temp = temp.next
            if temp == self.head:
                break
        print("(Back to Head)")

# --- Testing the Code ---
obj = DCll()
obj.insert(10)
obj.insert(20)
obj.insert(30)
obj.display()

print("\nDeleting 20...")
obj.delete(20)
obj.display()

print("\nDeleting 10 (Head)...")
obj.delete(10)
obj.display()
```
