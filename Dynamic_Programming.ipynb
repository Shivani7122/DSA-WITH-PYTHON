### ***WHY WE USE DYNAMIC PROGRAMMING?***

---
As we know that when the time complexity of program is in exponential it means, 2 to power n,etc. Then we use dynamic programming to convert the time complexity of the code from

          EXPONENTIAL ----->  POLYNOMIAL
Screenshot 2026-04-09 205342.png


Here we don't call any function again and again ,just simply store it's value,which reduces the time complexity.


Now we'll solve the fibonacci series problem using Dynamic Prgramming.

Normally,It's T(N) = O($2^n$)

After applying DP, T(n) = O(n)

```python
a = []
n = int(input("Enter the n : "))
for i in range(n+1):
  if i == 0 or i == 1:
    a.append(1)
  else:
    a.append(a[i-1]+a[i-2])
print(f"Fibonacci series : {a}")

```
