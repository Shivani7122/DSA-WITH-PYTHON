```python
def fractionalKnapsack(item_wt, price , capacity):
  n = len(item_wt)

  items = [(price[i],item_wt[i] , price[i]//item_wt[i]) for i in range(n)]

  for i in range(n):
    for j in range(i+1,n):
      if items[i][2] < items[j][2] :
        items[i],items[j] = items[j],items[i]



  profit = 0.0

  for price, item_wt,per_kg_price in items:
    if capacity >= item_wt:
      capacity -= item_wt
      profit += price
    else:
      profit += per_kg_price * capacity
      break
  print(f"Total profit = {profit}")


price = [24,21,12,10]
item_wt = [7,3,4,5]
capacity = 20
fractionalKnapsack(item_wt,price,capacity)


```
