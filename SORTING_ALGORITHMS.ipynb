# **BUBBLE SORT**


---
Comparison between the consecutive two elements.

Time Complexity:- O(n²)


```python
def bubbleSort(a,len):
  n = len

  for i in range(n):
    for j in range(0,n-1-i):
      if a[j] > a[j+1]:
        a[j],a[j+1] = a[j+1],a[j]

# a = [64,32,25,45,51,2]  -----> direct input
a = []
len = int(input("Enter the length of the list"))
for x in range(0,len):
  a.append(int(input("Enter the list item")))
print(a)
bubbleSort(a,len)
print(f"Sorted array --> {a}")

```

# **SELECTION SORT**
---
Minimum element ko sbse compare kerke jo usse less hoga usse swap ker dena h .


Time Complexity = O(n²)



```python
def selectionSort(a,len):
  n = len
  for i in range(n):
    min = i
    for j in range(i,n):
      if a[min] > a[j]:
        min = j
    a[i],a[min] = a[min],a[i]
a = []
len = int(input("Enter the length of the list "))
for x in range(0,len):
  a.append(int(input("Enter the list item ")))
print(a)
selectionSort(a,len)
print(f"Sorted array --> {a}")
```

# **INSERTION SORT**

---

Key ek point kerke uske left me compare kerenge aur wha usse bda koi hoga toh shift krenge.

Time Complexity: O(n²)

```python
def insertionSort(a,len):
  n = len
  for i in range(1,n):
    key = a[i]
    j = i-1
    while (j >= 0 and key < a[j]):
      a[j+1] = a[j]
      j -= 1
    a[j+1] = key
a = []
len = int(input("Enter the length of the list "))
for x in range(0,len):
  a.append(int(input("Enter the list item ")))
print(a)
insertionSort(a,len)
print(f"Sorted array --> {a}")

```

# **MERGE SORT**

---
It's an example of **Divide and Conquer**.

Time Complexity: O(n logn)


```python
def Divide(arr,l,r):
  if l < r:
    m = (l + r) // 2
    Divide(arr , l , m)
    Divide(arr , m+1 , r)
    merge(arr , l , m , r)

def merge(arr , l , m , r):
  s1 = m - l + 1
  s2 = r - m    # r - (m+1)+1 = r - m -1 +1 = r - m

  L = [0] * s1   #size of L array.
  R = [0] * s2   #size of R array.


#Copying from original array to L.
  for i in range(s1):
    L[i] = arr[l+i]

#Copying from original array to R.
  for j in range(s2):
    R[j] = arr[m+1+j]

  i = j = 0
  k = l   #because our array will start from l.

#Copying sorted elements from L and R to the original array back.
  while (i < s1 and j < s2):
    if L[i] <= R[j]:
      arr[k] = L[i]
      i += 1
    else:
      arr[k] = R[j]
      j += 1
    k += 1


#for remaining left elements.
  while (i < s1):
    arr[k] = L[i]
    i += 1
    k += 1

  while (j < s2):
    arr[k] = R[j]
    j += 1
    k += 1

def mergeSort(arr,len):
  n = len
  Divide(arr , 0 , n-1)

arr = []
len = int(input("Enter the length of the list "))
for x in range(0,len):
  arr.append(int(input("Enter the list item ")))
print(arr)
mergeSort(arr,len)
print(f"Sorted array --> {arr}")
```

# **QUICK SORT**


---


*   Based on Divide and Conquer.
*   Here we will select a pivot element and place it at its right place in the array.
*   Now the array need not be sorted it should be s.t the left of pivot should have elements less than it and right side elements should be greater than the pivot.


Time Complexity: O(n logn)

Worst Case Complexity: O(n2)




```python
def quickSort(arr, l , r):
  if l < r:
    p = partition(arr , l , r)

    quickSort(arr , l , p - 1)
    quickSort(arr , p + 1 , r)

def partition(arr , l , r):
  pivot = arr[l]
  i = l + 1
  j = r

  while True:
    while (i <= j and arr[i] < pivot):
      i += 1
    while (i <= j and arr[j] > pivot):
      j -= 1

    if i <= j:
      arr[i], arr[j] = arr[j], arr[i]
    else:
      break

  arr[l], arr[j] = arr[j], arr[l]
  return j
arr = []
n = int(input("Enter the length of the list "))
for x in range(0,n):
  arr.append(int(input("Enter the list item ")))
print(arr)
quickSort(arr , 0 , len(arr)-1)
print(f"Sorted array --> {arr}")



```

*NOTE:    Merge Sort is better than Quick Sort because in Merge sort the time complexity is always O(n logn), whereas in quick sort it depends on the pivot element i.e. totally based on the luck!!  :)*
