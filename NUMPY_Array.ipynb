ARITHMENTIC PROGRESSION ON ARRAY

```python
#Partition------------ divide the number in specified parts
ab = linspace(10,20,3)
for a in ab:
  print(a,end =" ")
```

```python
#arange()-----------common difference ap
au = arange(10,20,3)
for x in au:
  print(x, end = " ")
```

```python
#logspace()-----commom diff is in log
au = logspace(10,20,2)
for x in au:
  print(x, end = " ")
```

```python
#zeros()-------make array of zeroes
val = zeros(5)
for s in val:
  print(s,end=" ")
```

```python
#ones()-------make array of one
val = ones(6)
for s in val:
  print(s, end = " ")
```

```python
#full()-----------for numbers other than 0's or 1's

val = full(5,4)
for s in val:
  print(s, end = " ")
```
