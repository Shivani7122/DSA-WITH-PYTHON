**Graph Representation :-**
   
   1. Matrix
   2. List







**1)  Matrix Representation:**


Rows --> horizontal --> source --> Left to right

columns --> vertical --> destination --> top to bottom


In matrices, we notate the matrix with m x n ----> where,

    m: no. of rows
    n: no of columns


```python
class Graph:
  def __init__(self, vertex):
    self.mat = [[0]*vertex for x in range(vertex)]
    self.size = vertex

  def addEdge(self,src,dest):  #for weighted graph we will take weight parameter and instead of 1 we will write weight there
    if( 0 <= src < self.size and 0 <= dest < self.size):
      self.mat[src][dest] = 1
      self.mat[dest][src] = 1  #only for undirected graph
    else:
      print("Invalid Edge")

  def removeEdge(self,src,dest):
    self.mat[src][dest] = 0
    self.mat[dest][src] = 0 #only for undirected graph

  # map(str, row) ---> this change the row to a single string and then we join with space
  def print(self):
    for row in self.mat:
      print(' '.join(map(str,row)))

G = Graph(5)
G.addEdge(0,1)
G.addEdge(0,2)
G.addEdge(1,3)
G.addEdge(2,4)
G.addEdge(2,3)
G.addEdge(3,4)
G.removeEdge(0,1)
G.print()
```

* ADVANTAGES:- Simple and easy to understand code

* DISADVANTAGES:- Memory wastage , no point to write 0 repeatedly .

* DENSE GRAPH --> if no. of 1's > 0's.

* SPARSE GRAPH --> if no. of 0's > 1's.

**2) List Representation:**


---

  ADJACENCY LIST:

```python
class Graph:
  def __init__(self,vertex):
    self.adjLst = {}

  def addvertex(self,vertex):
    if vertex not in self.adjLst.keys():
      self.adjLst[vertex] = []

  def addEdge(self,src,dest):
    self.addvertex(src)   #only for directed graph
    self.addvertex(dest)

    self.adjLst[src].append(dest)
    self.adjLst[dest].append(src)

  def removeEdge(self,src,dest):
    self.adjLst[src].remove(dest)
    self.adjLst[dest].remove(src)

  def print(self):
    for vertex in self.adjLst:
      print(vertex, " --> ", self.adjLst[vertex], end = "\n")

G = Graph(6)
G.addEdge(1,2)
G.addEdge(2,3)
G.addEdge(1,4)
G.addEdge(2,4)
G.addEdge(3,4)
G.addEdge(3,5)
G.addEdge(4,5)
G.print()
```

# **GRAPH TRAVERSAL**

## **1. Depth First Search (DFS)**

```python
class Graph:
  def __init__(self, vertex):
    self.mat = [[0]*vertex for x in range(vertex)]
    self.size = vertex

  def addEdge(self,src,dest):  #for weighted graph we will take weight parameter and instead of 1 we will write weight there
    if( 0 <= src < self.size and 0 <= dest < self.size):
      self.mat[src][dest] = 1
      self.mat[dest][src] = 1  #only for undirected graph
    else:
      print("Invalid Edge")

  def removeEdge(self,src,dest):
    self.mat[src][dest] = 0
    self.mat[dest][src] = 0 #only for undirected graph

  # map(str, row) ---> this change the row to a single string and then we join with space
  def print(self):
    for row in self.mat:
      print(' '.join(map(str,row)))

  def DFS(self,src):
    visited = [False]*self.size
    stack = [src]

    while stack :
      v = stack.pop()

      if visited[v] == False:
        print(v, end = " -> ")
        visited[v] = True

      for i in range(self.size):
        if self.mat[v][i]== 1 and visited[i] == False:
          stack.append(i)

g = Graph(6)
g.addEdge(0,1)
g.addEdge(0,2)
g.addEdge(2,3)
g.addEdge(2,4)
g.addEdge(4,5)
g.addEdge(3,5)
g.print()
g.DFS(0)
```

## **2. Breadth First Search (BFS)**

```python
from collections import deque
class Graph:
  def __init__(self, vertex):
    self.mat = [[0]*vertex for x in range(vertex)]
    self.size = vertex

  def addEdge(self,src,dest):  #for weighted graph we will take weight parameter and instead of 1 we will write weight there
    if( 0 <= src < self.size and 0 <= dest < self.size):
      self.mat[src][dest] = 1
      self.mat[dest][src] = 1  #only for undirected graph
    else:
      print("Invalid Edge")

  def removeEdge(self,src,dest):
    self.mat[src][dest] = 0
    self.mat[dest][src] = 0 #only for undirected graph

  # map(str, row) ---> this change the row to a single string and then we join with space
  def print(self):
    for row in self.mat:
      print(' '.join(map(str,row)))

  def BFS(self,src):
    visited = [False]*self.size
    queue = deque([src])
    visited[src] = True

    while queue :
      v = queue.popleft()    #popleft()---> removes the element from the left or from start(front)
      print(v, end = " -> ")

      for i in range(self.size):
        if self.mat[v][i]== 1 and visited[i] == False:
          visited[i] = True
          queue.append(i)


G = Graph(8)
G.addEdge(0,1)
G.addEdge(0,3)
G.addEdge(1,3)
G.addEdge(3,5)
G.addEdge(3,4)
G.addEdge(4,5)
G.addEdge(4,6)
G.addEdge(6,2)
G.addEdge(6,7)
G.BFS(0)
```
