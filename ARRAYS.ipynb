```python
val = array.array('i',[1,2,3,4,5,6])
print(val)
```

```python
#Directly import array

from array import *
val = array('i',[1,2,3,4,5,6])
for x in val :
  print(x, end=" ")
```

```python
val.insert(3,50)
for x in val:
  print(x,end=" ")
```

```python
#Inserting values at the end of array
val.append(1)
for x in val:
  print(x,end=" ")
```
