### ***RECURSION***

Function calling itself is called ***Recursion***



1.   a = 0 ----> BASE CASE (Start of the problem)

2.   logic -----> the main code

3.   terminating contion ---->end of loop



### **1. FACTORIAL OF A NUMBER:-**

LOOP CODE

```python
num = int(input("Enter the number: "))
fact = 1
for i in range(1,num+1):
  fact *= i
print(f"Factorial of {num} is {fact} ")
```

RECURSIVE CODE

```python
def factorial(n):
  if n == 0 or n == 1:    #---- terminating condition
    return 1
  return n * factorial(n-1)   #----Recursion

n = int(input("Enter the number: "))
print(factorial(n))
```

# **2. FIBONACCI SERIES**

LOOP METHOD

```python
num = int(input ("Enter the length of fibonnaci series: "))
a = 1
b = 1
for i in range(num):
  print(a,end = " ")
  a,b = b,a + b
```

RECURSIVE METHOD


```python
def fib(n):
  if n == 1 or n == 2:
    return 1
  else:
    return fib(n-1) + fib(n-2)

n = int(input("Enter the length of fibonacci series: "))
for i in range(1,n+1):
  print(fib(i),end = " ")
```
