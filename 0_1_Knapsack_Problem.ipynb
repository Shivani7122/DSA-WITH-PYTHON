***EXPONENTIAL T(n) CODE SOLUTION:-***

```python
def knapsackRecusrion(weights , values , capacity , n):
  if n == 0 or capacity == 0:
    return 0
  if weights[n-1] > capacity:
    return knapsackRecusrion(weights , values , capacity , n-1)

  #if picked the item
  include = values[n-1] + knapsackRecusrion(weights , values , capacity - weights[n-1] , n-1)

  #If leave the item
  exclude = knapsackRecusrion(weights , values , capacity , n-1)

  #will return the max of the both the cases
  return max(include , exclude)

weights = [10 , 20 , 30]
values = [60 , 100 , 120]
capacity = int(input("Enter the capacity : "))
n = len(weights)
print("Maximum profit : ",knapsackRecusrion(weights , values , capacity , n))
```

```python
def knapsackDP(wt , value , capacity):
  n = len(wt)

  dp = [[0]*(capacity+1)for _ in range(n+1)]        #creating row of 0's.

  for i in range(n+1) :
    for w in range(capacity+1):
      #if we can keep the item in knapsack
      if (wt[i-1] <= w):
        dp[i][w] = max(dp[i-1][w] , dp[i-1][w - wt[i-1]] + value[i-1])
      else:
        dp[i][w] = dp[i-1][w]

  print(f"Maximum profit : {dp[n][capacity]}")


weights = [2,3,4,5]
values = [3,4,5,6]
capacity = int(input("Enter capacity : "))

knapsackDP(weights,values,capacity)
```
