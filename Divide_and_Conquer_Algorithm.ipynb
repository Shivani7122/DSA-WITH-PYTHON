Ques-1. Finding MIN and MAX .

```python
def find_min_max(arr , start , end):

  #BASE CASE

  #If there is only one element
  if start == end:
    return arr[start] , arr[start]

  #If there are two elements
  if start + 1 == end:
    if arr[start] < arr[end]:
      return arr[start] , arr[end]
    else:                #-------> arr[start] > arr[end]
      return arr[end] , arr[start]

  #If there are more than two elements
  mid = (start + end) // 2
  #In PYTHON we can assign/return more than 1 value in a single line .
  min1 , max1 = find_min_max(arr , start , mid)
  min2 , max2 = find_min_max(arr , mid + 1 , end)

  return min(min1 , min2) , max(max1 , max2)

#Function Call
arr = []
n = int(input("Enter the size of array : "))
for i in range(n):
  arr.append(int(input("Enter the element : ")))
print(arr)
min,max = find_min_max(arr, 0 , len(arr)-1)
print(f"Minimum element is : {min}")
print(f"Maximum element is : {max}")
```
